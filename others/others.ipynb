# addition with broadcasting

```python
import numpy as np

# random input
a = np.random.randn(100, 2)
b = np.random.randn(100, 2)


# pairwise addition grid, with broadcasting ([100, 1, 2] + [1, 100, 2])
m =  a[:, None, :] + b[None, :, :]

print("a shape:", a.shape)      # (100, 2)
print("b shape:", b.shape)      # (100, 2)
print("m shape:", m.shape)      # (100, 100, 2)

# verify one element
i, j = 3, 7
print("m[i, j]:", m[i, j])
print("a[i] + b[j]:", a[i] + b[j])
```
